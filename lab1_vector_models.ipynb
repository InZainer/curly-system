# Лабораторная работа 1 (бонус): Построение векторной модели текста

**Дисциплина:** NLP  
**Язык:** Python (Jupyter Notebook)

В этой работе для заданного текстового корпуса построены три векторные модели:
1. **One-hot encoding** (прямое кодирование)
2. **Bag-of-Words (BoW)**
3. **TF-IDF**


```python
# Импорты
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
# Текстовый корпус (можно заменить на свой)
corpus = [
    "Машинное обучение помогает анализировать текстовые данные",
    "Обработка естественного языка включает токенизацию и лемматизацию",
    "Векторные модели текста применяются в задачах NLP",
    "TF IDF уменьшает влияние часто встречающихся слов",
    "Мешок слов и one hot encoding являются базовыми подходами",
]

for i, doc in enumerate(corpus, 1):
    print(f"Документ {i}: {doc}")
```

## 1) One-hot encoding

Построим бинарное представление: 1 — термин присутствует в документе, 0 — отсутствует.


```python
one_hot_vectorizer = CountVectorizer(binary=True)
one_hot_matrix = one_hot_vectorizer.fit_transform(corpus)

one_hot_df = pd.DataFrame(
    one_hot_matrix.toarray(),
    columns=one_hot_vectorizer.get_feature_names_out(),
    index=[f"doc_{i}" for i in range(1, len(corpus)+1)]
)

print("Размерность матрицы One-hot:", one_hot_matrix.shape)
one_hot_df.head()
```

## 2) Bag-of-Words (BoW)

Построим частотное представление: вектор хранит **абсолютные частоты** слов в документе.


```python
bow_vectorizer = CountVectorizer(binary=False)
bow_matrix = bow_vectorizer.fit_transform(corpus)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_vectorizer.get_feature_names_out(),
    index=[f"doc_{i}" for i in range(1, len(corpus)+1)]
)

print("Размерность матрицы BoW:", bow_matrix.shape)
bow_df.head()
```

## 3) TF-IDF

Построим взвешенное представление, где вес термина отражает его важность в документе с учетом корпуса.


```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=[f"doc_{i}" for i in range(1, len(corpus)+1)]
)

print("Размерность матрицы TF-IDF:", tfidf_matrix.shape)
tfidf_df.round(3).head()
```

## Вывод

Для одного и того же корпуса получены три типа векторного представления:
- **One-hot** — бинарная фиксация наличия слова в документе;
- **BoW** — абсолютные частоты слов;
- **TF-IDF** — взвешенные частоты, уменьшающие вклад общеупотребительных слов.

Таким образом, требования лабораторной работы выполнены: реализованы и представлены все три модели векторизации текста.
